# Making our plots more appealing

```python
import numpy as np
import matplotlib.pyplot as plt

# Create some dummy data:


A = np.random.random(10)*10
B = np.random.random(10)*15
```

```python
#plt.bar([0, 1], [A,B])
fig = plt.figure(1, figsize=(5,5))
plt.plot(A)
plt.plot(B)
```

```python
fig = plt.figure(1, figsize=(5,5))
plt.plot(A, lw=3, c="k", linestyle="-")
plt.plot(B, lw=3, c="grey", linestyle="-.")

ax = plt.gca() # get current axis
ax.spines["right"].set_visible(False)
#ax.spines["left"].set_visible(False)
ax.spines["top"].set_visible(False)
#ax.spines["bottom"].set_visible(False)
ax.spines["bottom"].set_linewidth(3)
ax.spines["left"].set_linewidth(3)

my_xticks = np.arange(0,10,1)
my_yticks = np.arange(0,16,5)
plt.xticks(my_xticks, fontsize=16)
plt.yticks(my_yticks, fontsize=16)
```

```python
# and now color-blind save:
fig = plt.figure(1, figsize=(5,5))

#plt.grid(color='gainsboro', linestyle='-', linewidth=0.25, 
#         zorder=1)

plt.plot(A, lw=3, c="#E69F00", linestyle="-")
plt.plot(B, lw=3, c="#56B4E9", linestyle="-.")

ax = plt.gca() # get current axis
ax.spines["right"].set_visible(False)
#ax.spines["left"].set_visible(False)
ax.spines["top"].set_visible(False)
#ax.spines["bottom"].set_visible(False)
ax.spines["bottom"].set_linewidth(3)
ax.spines["left"].set_linewidth(3)

ax.xaxis.set_tick_params(width=3, length=8)
ax.yaxis.set_tick_params(width=3, length=8)

my_xticks = np.arange(0,10,1)
my_yticks = np.arange(0,16,5)
plt.xticks(my_xticks, fontsize=16)
plt.yticks(my_yticks, fontsize=16)
```
